How it works

"I ate biryani for lunch today."

The system saves that as a memory.

Later:

"What did I eat last week?"

The system retrieves only the relevant memories and gives them to Qwen to generate the answer.

```python
import ollama

response = ollama.chat(
    model='llama3.1',
    messages=[
        {
            'role': 'user',
            'content': 'Explain machine learning in simple terms.'
        }
    ]
)

print(response['message']['content'])


```

```python
import ollama

response = ollama.chat(
    model="qwen3:1.7b",
    messages=[
        {
            "role": "user",
            "content": "Hello, explain machine learning in simple terms."
        }
    ]
)

print(response.message.content)
```
