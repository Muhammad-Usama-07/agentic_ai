```python
import requests

url = "http://localhost:11434/api/chat"

payload = {
    "model": "qwen3:1.7b",
    "messages": [
        {
            "role": "user",
            "content": "Explain machine learning in simple terms."
        }
    ],
    "think": False,
    "stream": False
}

response = requests.post(url, json=payload)

print(response.json()["message"]["content"])
```
